```python
data.groupby("Country")["Category"].count()
```

```python
data.groupby("Category")["Country"].count()
```

```python
data.groupby(["Country","Category"]).size().reset_index(name="Count")
```

```python
import matplotlib.pyplot as plt

result = data.groupby(["Country", "Category"]).size().reset_index(name="Count")

pivot = result.pivot(index="Country", columns="Category", values="Count").fillna(0)
pivot.plot(kind="bar", figsize=(10, 6))

plt.title("Purchase Count by Country and Category")
plt.xlabel("Country")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
