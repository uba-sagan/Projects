## Concate Operation -- joining two or more dataframes 

```python
import pandas as pd

india_weather = pd.DataFrame({
    "city": ["mumbai","delhi","banglore"],
    "temperature": [32,45,30],
    "humidity": [80, 60, 78]
})
india_weather
```

```python
us_weather = pd.DataFrame({
    "city": ["new york","chicago","orlando"],
    "temperature": [21,14,35],
    "humidity": [68, 65, 75]
})
us_weather
```

```python
df_concat = pd.concat([india_weather,us_weather])

df_concat
```

```python
#Ignore the Index
df_concat = pd.concat([india_weather, us_weather], ignore_index=False)

df_concat
```

```python
df_concat_1 = pd.concat([india_weather, us_weather], ignore_index=True)

df_concat_1
```

```python
# Concatenation And Keys
df_concat_2 = pd.concat([india_weather, us_weather], keys=["india", "us"], )


df_concat_2
```

## types of database joins

```python
## Inner joins / Intersections : Get the common between the two elements
## Outer Join / Union :  all elements from both sets, whether they match or not

'''how="outer" → full outer join (union of keys).

how="left" → left outer join.

how="right" → right outer join.

how="inner" → intersection (only matching keys).'''
```

```python
df3=pd.merge(df1,df2,on="city",how="outer")
df3
```

```python
df3=pd.merge(df1,df2,on="city",how="left")
df3
```

```python
df3=pd.merge(df1,df2,on="city",how="right")
df3
```

```python
df3=pd.merge(df1,df2,on="city",how="right", indicator=True)
df3
```

```python
# suffixes
df1 = pd.DataFrame({
    "city": ["new york","chicago","orlando", "baltimore"],
    "temperature": [21,14,35,38],
    "humidity": [65,68,71, 75]
})

df1
```

```python
df3= pd.merge(df1,df2,on="city",how="outer", suffixes=('_first','_second'))
df3
```
